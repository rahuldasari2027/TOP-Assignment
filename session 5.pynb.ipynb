Task 1: Rebuild trips and calculate fares

```python

import numpy as np

np.random.seed(42)

marks = np.random.randint(35, 101, size=(40, 5))

np.random.seed(7)
retest_marks = marks + np.random.randn(40, 5) * 3

fee_inr = np.random.choice(
    np.linspace(1000, 5000, 5),
    size=40
)

arrays = {
    "marks": marks,
    "retest_marks": retest_marks,
    "fee_inr": fee_inr
}

print("Array Attributes")
print("-" * 85)
print(f"{'Array':<15} {'ndim':<6} {'shape':<12} {'size':<7} {'dtype':<10} {'itemsize':<10} {'nbytes':<10}")

for name, arr in arrays.items():
    print(f"{name:<15} {arr.ndim:<6} {str(arr.shape):<12} {arr.size:<7} {str(arr.dtype):<10} {arr.itemsize:<10} {arr.nbytes:<10}")
    assert arr.nbytes == arr.size * arr.itemsize

print("All attribute checks passed!")
```

Task 2: Logarithms and square roots

```python

import numpy as np

np.random.seed(42)

trip_id = np.arange(1, 501)
distance_km = np.random.rand(500) * 24 + 1
wait_min = np.random.randint(0, 16, 500)
hour = np.random.randint(6, 24, 500)
surge_level = np.random.randint(0, 4, 500)

fare_fn = np.multiply(
    np.add(
        np.add(20, np.multiply(12, distance_km)),
        np.multiply(1.5, wait_min)
    ),
    np.power(1.1, surge_level)
)

print("Task 1 completed")
print("Total Trips:", len(trip_id))
print("First 5 Fares:", fare_fn[:5])
```

Task 3: Calculate straight-line distance using the Haversine formula

```python

np.random.seed(5)

pickup_lat = 22.95 + 0.15 * np.random.rand(10)
pickup_lon = 72.50 + 0.15 * np.random.rand(10)

drop_lat = 22.95 + 0.15 * np.random.rand(10)
drop_lon = 72.50 + 0.15 * np.random.rand(10)

lat1 = np.radians(pickup_lat)
lon1 = np.radians(pickup_lon)

lat2 = np.radians(drop_lat)
lon2 = np.radians(drop_lon)

dlat = np.subtract(lat2, lat1)
dlon = np.subtract(lon2, lon1)

a = (
    np.sin(dlat / 2) ** 2
    + np.cos(lat1) * np.cos(lat2)
    * np.sin(dlon / 2) ** 2
)

distance = 2 * 6371 * np.arcsin(np.sqrt(a))

print("Straight-Line Distances (KM):")
print(np.round(distance, 2))

# Trigonometric identity checks
assert np.allclose(
    np.degrees(np.arcsin(0.5)), 30
)

assert np.allclose(
    np.tan(np.radians(45)), 1
)

print("Trigonometric checks passed!")
```

Task 4: Fare statistics and yearly hikes

```python

print("FARE STATISTICS")
print("--------------------------")

print("Mean Fare:", np.round(np.mean(fare_fn), 2))
print("Median Fare:", np.median(fare_fn))
print("Standard Deviation:", np.std(fare_fn, ddof=0))
print("Variance:", np.var(fare_fn))

print("Minimum Fare:", np.floor(np.min(fare_fn)))
print("Maximum Fare:", np.max(fare_fn))
print("Total Fare:", np.sum(fare_fn))

print("25th Percentile:", np.percentile(fare_fn, 25))
print("50th Percentile:", np.percentile(fare_fn, 50))
print("75th Percentile:", np.percentile(fare_fn, 75))
print("90th Percentile:", np.percentile(fare_fn, 90))

print("90th Percentile Rounded Up:",
      np.ceil(np.percentile(fare_fn, 90)))

# Cumulative revenue across 25 groups of 20 trips
group_revenue = fare_fn.reshape(25, 20).sum(axis=1)

cumulative_revenue = np.cumsum(group_revenue)

print("Cumulative Revenue:")
print(cumulative_revenue)

# Combined yearly fare increase
yearly_hikes = np.array([1.05, 1.10, 1.08])

combined_hike = np.prod(yearly_hikes)

print("Combined Hike Multiplier:", combined_hike)
print("Total Percentage Increase:",
      np.round((combined_hike - 1) * 100, 2))
```
