Matplotlib Session 3 – Bar Charts (Vertical + Horizontal)

```python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set random seed
np.random.seed(42)

# Society wings A to F
wings = ["A", "B", "C", "D", "E", "F"]

# Generate electricity usage data
df = pd.DataFrame({
    "wing": wings,
    "units_this_month": np.random.randint(
        8000, 20001, size=6
    ),
    "units_last_month": np.random.randint(
        8000, 20001, size=6
    )
})

# Save dataset as CSV
df.to_csv("society_power.csv", index=False)

# Display dataset
print("HOUSING SOCIETY ELECTRICITY DATA")
print(df)

print("\nDataset Shape:", df.shape)
print("\nCSV file created successfully!")
```

Task 2: Vertical Bar Chart

```python

# Matplotlib Session 3 - Task 2
# Vertical Bar Chart

# Sort electricity usage highest to lowest
df_sorted = df.sort_values(
    by="units_this_month",
    ascending=False
)

# Create vertical bar chart
plt.figure(figsize=(8, 5))

plt.bar(
    df_sorted["wing"],
    df_sorted["units_this_month"],
    color="steelblue",
    width=0.6
)

# Add title and axis labels
plt.title("Electricity Units by Wing")
plt.xlabel("Society Wing")
plt.ylabel("Electricity Usage (kWh)")

# Add values above each bar
for i, value in enumerate(df_sorted["units_this_month"]):
    plt.text(
        i,
        value + 200,
        str(value),
        ha="center"
    )

plt.ylim(0, 23000)
plt.tight_layout()

# Save BEFORE plt.show()
plt.savefig("society_power_bar.png")

# Display chart
plt.show()

print("Vertical bar chart saved successfully!")
```

Task 3: Grouped Bar Chart

```python

# Matplotlib Session 3 - Task 3
# Grouped Bar Chart

import numpy as np
import matplotlib.pyplot as plt

# Set positions for 6 wings
x = np.arange(6)

# Set bar width
width = 0.4

plt.figure(figsize=(9, 5))

# Last month bars - shifted left
plt.bar(
    x - 0.2,
    df["units_last_month"],
    width=width,
    color="orange",
    label="Last Month"
)

# This month bars - shifted right
plt.bar(
    x + 0.2,
    df["units_this_month"],
    width=width,
    color="steelblue",
    label="This Month"
)

# Set wing names on X-axis
plt.xticks(x, df["wing"])

# Chart formatting
plt.title("Electricity Usage: Last Month vs This Month")
plt.xlabel("Society Wing")
plt.ylabel("Electricity Usage (kWh)")
plt.legend()

plt.tight_layout()

# Save chart BEFORE plt.show()
plt.savefig("society_power_grouped.png")

# Display chart
plt.show()

print("Grouped bar chart saved successfully!")
```

Task 4: Horizontal Bar Chart

```python

# Matplotlib Session 3 - Task 4
# Horizontal Bar Chart

import matplotlib.pyplot as plt

# Sort electricity usage from highest to lowest
df_horizontal = df.sort_values(
    by="units_this_month",
    ascending=False
)

# Create horizontal bar chart
plt.figure(figsize=(9, 5))

plt.barh(
    df_horizontal["wing"],
    df_horizontal["units_this_month"],
    color="teal"
)

# Highest value at the top
plt.gca().invert_yaxis()

# Add kWh values at the end of each bar
for i, value in enumerate(df_horizontal["units_this_month"]):
    plt.text(
        value + 200,
        i,
        str(value),
        va="center"
    )

# Add chart title and labels
plt.title("Electricity Usage by Wing - Horizontal Bar Chart")
plt.xlabel("Electricity Usage (kWh)")
plt.ylabel("Society Wing")

plt.xlim(0, 23000)
plt.tight_layout()

# Save BEFORE plt.show()
plt.savefig("society_power_barh.png")

# Display chart
plt.show()

print("Horizontal bar chart saved successfully!")
```
