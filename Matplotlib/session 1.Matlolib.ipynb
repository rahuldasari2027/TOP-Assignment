Python Matplotlib – Session 1

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Kirana Store Sales - 30 Days
np.random.seed(42)

df = pd.DataFrame({
    "day": range(1, 31),
    "sales_inr": np.random.randint(4000, 12001, size=30),
    "customers": np.random.randint(60, 221, size=30)
})

# Save dataset
df.to_csv("kirana_sales_30d.csv", index=False)

# Display dataset
print(df)

print("\nDataset Shape:", df.shape)
print("\nCSV file created successfully!")
```

Task 2: Create a Sales Line Chart

```python


print("SALES SUMMARY")
print(df.describe())

# Find the 3 highest-sales days
top_3 = df.nlargest(3, "sales_inr")

print("\nTOP 3 HIGHEST SALES DAYS")
print(top_3[["day", "sales_inr"]])

# Create line chart
plt.figure(figsize=(8, 4))

plt.plot(
    df["day"],
    df["sales_inr"],
    marker="o",
    color="blue"
)

plt.title("30 Days Kirana Store Sales")
plt.xlabel("Day")
plt.ylabel("Sales (INR)")

plt.show()
```

Task 3: Figure and Axes Objects

```python



import matplotlib.pyplot as plt

# Create Figure
fig = plt.figure(figsize=(8, 4))

# Add Axes
ax = fig.add_axes([0.1, 0.15, 0.85, 0.75])

# Plot sales against day
ax.plot(
    df["day"],
    df["sales_inr"],
    marker="o",
    color="blue"
)

# Set axis limits
ax.set_xlim(1, 30)
ax.set_ylim(0, 14000)

# Add title and labels
ax.set_title("30 Days Kirana Store Sales")
ax.set_xlabel("Day")
ax.set_ylabel("Sales (INR)")

# Print object types
print("Figure Type:", type(fig))
print("Axes Type:", type(ax))
print("Axis Type:", type(ax.xaxis))

# Save chart BEFORE plt.show()
plt.savefig("kirana_sales_basic.png")

# Display chart
plt.show()

print("Chart saved successfully!")
```
