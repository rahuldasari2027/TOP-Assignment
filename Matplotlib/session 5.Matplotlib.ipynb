Matplotlib Session 5 – Histogram (Distribution Plot)

```python

# Matplotlib Session 5 - Task 1
# Generate Coaching Class Student Data

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set random seed
np.random.seed(42)

# Generate 300 students
n = 300

df = pd.DataFrame({
    "student_id": range(1, n + 1),

    "batch": np.random.choice(
        ["Morning", "Evening"], size=n
    ),

    "maths_marks": np.clip(
        np.random.normal(62, 14, size=n),
        0, 100
    ),

    "attendance_pct": np.random.randint(
        55, 101, size=n
    )
})

# Round marks to 2 decimal places
df["maths_marks"] = df["maths_marks"].round(2)

# Save CSV file
df.to_csv("coaching_marks.csv", index=False)

# Display data
print("COACHING CLASS STUDENT DATA")
print(df.head(10))

print("\nDataset Shape:", df.shape)
print("\nCSV file created successfully!")
```

Task 2: Create a Histogram

```python

# Matplotlib Session 5 - Task 2
# Histogram of Maths Marks

import matplotlib.pyplot as plt

# Create histogram
plt.figure(figsize=(8, 5))

plt.hist(
    df["maths_marks"],
    bins=10,
    edgecolor="black",
    color="skyblue"
)

# Add title and labels
plt.title("Maths Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of students")

plt.tight_layout()

# Save chart BEFORE plt.show()
plt.savefig("marks_hist_10.png")

# Display histogram
plt.show()

print("Task 2 completed successfully!")
print("marks_hist_10.png saved successfully!")
```

Task 3: Create Histograms with 5, 10 and 25 Bins

```python

# Matplotlib Session 5 - Task 3
# Compare Histograms with Different Bins

import matplotlib.pyplot as plt

# Create histograms with 5, 10 and 25 bins
for bins in [5, 10, 25]:

    # Create a new figure for each histogram
    plt.figure(figsize=(8, 5))

    plt.hist(
        df["maths_marks"],
        bins=bins,
        edgecolor="black",
        color="skyblue"
    )

    # Add title and labels
    plt.title(f"Maths Marks Distribution - {bins} Bins")
    plt.xlabel("Marks")
    plt.ylabel("Number of students")

    plt.tight_layout()

    # Save BEFORE plt.show()
    plt.savefig(f"marks_bins_{bins}.png")

    plt.show()

    print(f"marks_bins_{bins}.png saved successfully!")

print("\nTask 3 completed successfully!")
```

Task 4: Density Histogram

```python

# Matplotlib Session 5 - Task 4
# Density Histogram

import numpy as np
import matplotlib.pyplot as plt

# Create density histogram
plt.figure(figsize=(8, 5))

# Capture histogram return values
counts, edges, patches = plt.hist(
    df["maths_marks"],
    bins=15,
    density=True,
    edgecolor="black",
    color="skyblue"
)

# Add title and labels
plt.title("Maths Marks Density Distribution")
plt.xlabel("Marks")
plt.ylabel("Density")

# Calculate total histogram area
total_area = (counts * np.diff(edges)).sum()

print("Total Histogram Area:", total_area)

# Verify area is approximately 1
print("Area approximately 1:", np.isclose(total_area, 1.0))

plt.tight_layout()

# Save BEFORE plt.show()
plt.savefig("marks_density.png")

plt.show()

print("Task 4 completed successfully!")
print("marks_density.png saved successfully!")
```
