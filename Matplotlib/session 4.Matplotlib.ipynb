Matplotlib Session 4 – Pie Chart & Donut Chart

```python

# Matplotlib Session 4 - Task 1
# Family Monthly Budget Pie Chart

import pandas as pd
import matplotlib.pyplot as plt

# Create household budget dataset
df = pd.DataFrame({
    "category": [
        "Rent",
        "Groceries",
        "Education",
        "Fuel",
        "Mobile and Internet",
        "Medical",
        "Entertainment",
        "Savings"
    ],
    "amount": [
        12000,
        9000,
        7000,
        3500,
        1800,
        2200,
        2500,
        2000
    ]
})

# Display budget data
print("FAMILY MONTHLY BUDGET")
print(df)

print("\nTotal Budget: Rs", df["amount"].sum())

# Create pie chart
plt.figure(figsize=(8, 6))

plt.pie(
    df["amount"],
    labels=df["category"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Family Monthly Budget - Rs 40,000")

plt.axis("equal")
plt.tight_layout()

# Save chart
plt.savefig("budget_pie.png")

plt.show()

print("Task 1 completed successfully!")
print("budget_pie.png saved successfully!")
```

Task 2: Explode the Largest Pie Slice

```python

# Matplotlib Session 4 - Task 2
# Explode the largest budget category

import matplotlib.pyplot as plt

# Find the largest category using idxmax()
largest_index = df["amount"].idxmax()
largest_category = df.loc[largest_index, "category"]

print("Largest Category:", largest_category)

# Build explode list using list comprehension
explode = [
    0.1 if category == largest_category else 0
    for category in df["category"]
]

print("Explode Values:", explode)

# Calculate total budget
total_budget = df["amount"].sum()

# Create pie chart
plt.figure(figsize=(8, 6))

plt.pie(
    df["amount"],
    labels=df["category"],
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    shadow=False
)

plt.title(f"Family Monthly Budget - Rs {total_budget:,}")

plt.axis("equal")
plt.tight_layout()

# Save chart BEFORE showing it
plt.savefig("budget_pie_explode.png")

plt.show()

print("Task 2 completed successfully!")
print("budget_pie_explode.png saved successfully!")
```

Task 3: Create a Donut Chart

```python

# Matplotlib Session 4 - Task 3
# Family Budget Donut Chart

import matplotlib.pyplot as plt

# Calculate total budget
total_budget = df["amount"].sum()

# Create donut chart
plt.figure(figsize=(8, 6))

plt.pie(
    df["amount"],
    labels=df["category"],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.4)
)

# Display total budget in the centre
plt.text(
    0, 0,
    f"Rs {total_budget:,}",
    ha="center",
    va="center",
    fontsize=16,
    fontweight="bold"
)

plt.title("Family Monthly Budget - Donut Chart")
plt.axis("equal")
plt.tight_layout()

# Save BEFORE plt.show()
plt.savefig("budget_donut.png")

plt.show()

print("Task 3 completed successfully!")
print("budget_donut.png saved successfully!")
```

Task 4: Group Small Categories into "Other"

```python

# Matplotlib Session 4 - Task 4
# Group categories below 6% into Other

import pandas as pd
import matplotlib.pyplot as plt

# Calculate total budget
total_budget = df["amount"].sum()

# Calculate percentage share
df["share"] = df["amount"] / total_budget * 100

# Find categories below 6%
mask = df["share"] < 6

print("CATEGORIES BELOW 6%")
print(df.loc[mask, ["category", "amount", "share"]])

# Rename small categories as Other
df_grouped = df.copy()
df_grouped.loc[mask, "category"] = "Other"

# Group amounts by category
df_grouped = (
    df_grouped.groupby("category", as_index=False)["amount"]
    .sum()
)

print("\nGROUPED BUDGET DATA")
print(df_grouped)

# Create grouped donut chart
plt.figure(figsize=(9, 6))

plt.pie(
    df_grouped["amount"],
    labels=df_grouped["category"],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.4)
)

# Display total budget in centre
plt.text(
    0, 0,
    f"Rs {total_budget:,}",
    ha="center",
    va="center",
    fontsize=16,
    fontweight="bold"
)

plt.title("Family Monthly Budget - Grouped Donut Chart")
plt.axis("equal")
plt.tight_layout()

# Save before showing
plt.savefig("budget_donut_grouped.png")

plt.show()

print("Task 4 completed successfully!")
print("budget_donut_grouped.png saved successfully!")
```
