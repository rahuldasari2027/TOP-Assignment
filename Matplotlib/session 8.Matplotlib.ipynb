Matplotlib Session 8 – Charts Customization & Style Management

```python

# MATPLOTLIB SESSION 8 - TASK 1
# Corporate Gym Membership Performance Chart

import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Load existing gym dataset
data_path = Path("gym_members.csv")

if not data_path.exists():
    raise FileNotFoundError(
        "gym_members.csv not found. "
        "Place the CSV in your notebook's working folder."
    )

df = pd.read_csv(data_path)

# Validate required columns
required_columns = {"plan", "visits_per_month"}

missing_columns = required_columns - set(df.columns)

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# Define membership plans
plan_order = ["Basic", "Standard", "Premium"]

# Calculate average monthly visits
avg_visits = (
    df.groupby("plan")["visits_per_month"]
    .mean()
    .reindex(plan_order)
)

if avg_visits.isna().any():
    raise ValueError("One or more membership plans have no data.")

# Assignment brand colors
brand_colors = [
    "#0B3C5D",
    "#328CC1",
    "#D9B310"
]

# Create chart
fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    avg_visits.index,
    avg_visits.values,
    color=brand_colors,
    width=0.6,
    edgecolor="white",
    linewidth=1
)

# Add data labels
for bar in bars:
    value = bar.get_height()

    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.2,
        f"{value:.1f}",
        ha="center",
        va="bottom",
        fontsize=11,
        fontweight="bold"
    )

# Professional chart formatting
ax.set_title(
    "Average Monthly Gym Visits by Membership Plan",
    fontsize=14,
    fontweight="bold",
    pad=15
)

ax.set_xlabel(
    "Membership Plan",
    fontsize=11
)

ax.set_ylabel(
    "Average Visits per Month",
    fontsize=11
)

# Light grid lines on Y-axis only
ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.25
)

ax.set_axisbelow(True)

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.set_ylim(0, avg_visits.max() * 1.2)

fig.tight_layout()

# Save before displaying
fig.savefig(
    "gym_brand_bar.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# Business summary
print("GYM MEMBERSHIP PERFORMANCE REPORT")
print("-" * 40)

print("\nAverage Monthly Visits by Plan:")
print(avg_visits.round(2))

print("\nTotal Members:", len(df))
print("Chart saved: gym_brand_bar.png")
print("Session 8 - Task 1 completed successfully!")
```

Task 2: Professional Line Chart

```python

# MATPLOTLIB SESSION 8 - TASK 2
# Monthly Gym Membership Joins by Plan

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("gym_members.csv")

# Validate required columns
required = {"join_month", "plan"}
missing = required - set(df.columns)

if missing:
    raise ValueError(f"Missing columns: {missing}")

# Define plans, brand colors and line styles
plans = ["Basic", "Standard", "Premium"]

colors = ["#0B3C5D", "#328CC1", "#D9B310"]
line_styles = ["-", "--", ":"]

# Calculate monthly joins for each plan
monthly_joins = (
    df.groupby(["join_month", "plan"])
    .size()
    .unstack(fill_value=0)
    .reindex(
        index=range(1, 13),
        columns=plans,
        fill_value=0
    )
)

# Create professional chart
fig, ax = plt.subplots(figsize=(10, 5))

for plan, color, style in zip(
    plans, colors, line_styles
):
    ax.plot(
        monthly_joins.index,
        monthly_joins[plan],
        linestyle=style,
        color=color,
        linewidth=2.5,
        marker="o",
        markersize=5,
        label=plan
    )

# Chart title and labels
ax.set_title(
    "Monthly Gym Membership Joins by Plan",
    fontsize=14,
    fontweight="bold",
    pad=15
)

ax.set_xlabel("Join Month")
ax.set_ylabel("Number of New Members")

ax.set_xticks(range(1, 13))

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.25
)

ax.legend(title="Membership Plan")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

fig.tight_layout()

# Save BEFORE displaying
fig.savefig(
    "gym_joins_linestyles.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# Print monthly membership report
print("MONTHLY GYM MEMBERSHIP REPORT")
print(monthly_joins)

print("\nTotal Members:", monthly_joins.to_numpy().sum())
print("Chart saved: gym_joins_linestyles.png")
```

Task 3: Compare Three Matplotlib Themes

```python

# MATPLOTLIB SESSION 8 - TASK 3
# Compare Professional Chart Themes

import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Load gym membership dataset
df = pd.read_csv("gym_members.csv")

# Validate required columns
if "join_month" not in df.columns:
    raise ValueError("Missing column: join_month")

# Calculate total monthly joins across all plans
monthly_joins = (
    df.groupby("join_month")
    .size()
    .reindex(range(1, 13), fill_value=0)
)

# Define chart styles and output filenames
chart_themes = {
    "ggplot": "gym_style_ggplot.png",
    "bmh": "gym_style_bmh.png",
    "dark_background": "gym_style_dark.png"
}

# Generate one chart for each theme
for style_name, filename in chart_themes.items():

    # Apply theme only within this block
    with plt.style.context(style_name):

        fig, ax = plt.subplots(figsize=(10, 5))

        # Plot monthly membership joins
        ax.plot(
            monthly_joins.index,
            monthly_joins.values,
            color="#328CC1",
            linestyle="-",
            linewidth=2.5,
            marker="o",
            markersize=6,
            label="New Members"
        )

        # Highlight the monthly values
        for month, count in monthly_joins.items():
            ax.annotate(
                str(count),
                xy=(month, count),
                xytext=(0, 8),
                textcoords="offset points",
                ha="center",
                fontsize=9
            )

        # Add chart title and axis labels
        ax.set_title(
            f"Monthly Gym Membership Joins - {style_name}",
            fontsize=14,
            fontweight="bold",
            pad=15
        )

        ax.set_xlabel("Month")
        ax.set_ylabel("Number of New Members")

        ax.set_xticks(range(1, 13))

        # Provide space for value labels
        ax.set_ylim(
            0,
            max(monthly_joins.max() * 1.2, 1)
        )

        ax.legend(loc="upper left")

        fig.tight_layout()

        # Save before displaying
        fig.savefig(
            filename,
            dpi=300,
            bbox_inches="tight"
        )

        plt.show()
        plt.close(fig)

        # Verify saved file
        file_path = Path(filename)

        if file_path.exists():
            print(
                f"Saved: {filename} "
                f"({file_path.stat().st_size / 1024:.1f} KB)"
            )

# Print all available Matplotlib styles
print("\nAVAILABLE MATPLOTLIB STYLES")
print(plt.style.available)

print("\nSESSION 8 - TASK 3 COMPLETED")
```
