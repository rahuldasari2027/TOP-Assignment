Matplotlib Session 2 – Basic Plotting (Line Charts)

```python


import pandas as pd
import matplotlib.pyplot as plt

# Load Session 1 dataset
df = pd.read_csv("kirana_sales_30d.csv")

print("DATASET PREVIEW")
print(df.head())

# Create daily sales line chart
plt.figure(figsize=(8, 4))

plt.plot(
    df["day"],
    df["sales_inr"],
    color="darkgreen",
    linewidth=2,
    marker="o",
    markersize=5
)

plt.title("Daily Sales - RK Kirana Store")
plt.xlabel("Day of month")
plt.ylabel("Sales (INR)")

plt.show()

print("Task 1 completed successfully!")
```

Task 2: Add a 7-Day Moving Average

```python

# Session 2 - Task 2
# Daily Sales vs 7-Day Moving Average

import matplotlib.pyplot as plt

# Calculate 7-day moving average
df["sales_7day_avg"] = df["sales_inr"].rolling(7).mean()

print("7-DAY MOVING AVERAGE")
print(df[["day", "sales_inr", "sales_7day_avg"]].head(10))

# Create line chart
plt.figure(figsize=(8, 4))

# Green daily sales line
plt.plot(
    df["day"],
    df["sales_inr"],
    color="darkgreen",
    linewidth=2,
    marker="o",
    markersize=5,
    label="Daily Sales"
)

# Orange moving average line
plt.plot(
    df["day"],
    df["sales_7day_avg"],
    color="orange",
    linewidth=3,
    label="7-Day Average"
)

# Chart formatting
plt.title("Daily sales (green) vs 7-day average (orange)")
plt.xlabel("Day of month")
plt.ylabel("Sales (INR)")
plt.legend()

plt.show()

print("Task 2 completed successfully!")
```

Task 3: Customer Line Chart and Save PNG Files

```python

# Session 2 - Task 3
# Customers Line Chart

plt.figure(figsize=(8, 4))

plt.plot(
    df["day"],
    df["customers"],
    marker="D",
    color="#C0392B",
    linewidth=1.5
)

plt.title("Daily Customers - RK Kirana Store")
plt.xlabel("Day of month")
plt.ylabel("Number of Customers")

plt.grid(True)
plt.tight_layout()

# Save BEFORE plt.show()
plt.savefig("kirana_customers_line.png")

plt.show()

print("Customer chart saved successfully!")
```

```python

# Save Sales Chart with 7-Day Moving Average

plt.figure(figsize=(8, 4))

plt.plot(
    df["day"],
    df["sales_inr"],
    color="darkgreen",
    linewidth=2,
    marker="o",
    markersize=5,
    label="Daily Sales"
)

plt.plot(
    df["day"],
    df["sales_7day_avg"],
    color="orange",
    linewidth=3,
    label="7-Day Average"
)

plt.title("Daily sales (green) vs 7-day average (orange)")
plt.xlabel("Day of month")
plt.ylabel("Sales (INR)")
plt.legend()
plt.tight_layout()

# Save BEFORE plt.show()
plt.savefig("kirana_sales_line.png")

plt.show()

print("Sales chart saved successfully!")
```
