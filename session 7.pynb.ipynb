# Housing Society Water Data – Array Manipulation
## NumPy Session 7 Assignment

Working with two housing society wings using NumPy stacking, concatenation, splitting, insertion, deletion, copying and views.

Task 1: Merge two society wings

```python

import numpy as np

# Rebuild Wing A data from Session 6
np.random.seed(42)
water_kl = np.random.randint(8, 41, size=(12, 20))

# Create Wing B data
np.random.seed(7)
water_b = np.random.randint(8, 41, size=(12, 15))

# Merge both wings
society_all = np.hstack((water_kl, water_b))

# Compare with concatenate
combined = np.concatenate((water_kl, water_b), axis=1)

assert society_all.shape == (12, 35)
assert np.array_equal(society_all, combined)

print("Wing A Shape:", water_kl.shape)
print("Wing B Shape:", water_b.shape)
print("Combined Society Shape:", society_all.shape)
print("HStack and Concatenate Match:", np.array_equal(society_all, combined))
```

Task 2: Compare stacking and concatenation

```python

np.random.seed(8)

electricity_kwh = np.random.randint(100, 401, size=(12, 20))

usage_rows = np.vstack((water_kl, electricity_kwh))

usage_cube = np.dstack((water_kl, electricity_kwh))

combined_rows = np.concatenate(
    (water_kl, electricity_kwh), axis=0
)

assert usage_rows.shape == (24, 20)
assert usage_cube.shape == (12, 20, 2)
assert np.array_equal(usage_rows, combined_rows)

print("Water Shape:", water_kl.shape)
print("Electricity Shape:", electricity_kwh.shape)
print("Vertical Stack Shape:", usage_rows.shape)
print("Depth Stack Shape:", usage_cube.shape)
print("VStack matches Concatenate:", np.array_equal(usage_rows, combined_rows))
```

Task 3: Split society data

```python

# Split 35 flats into 5 equal blocks
equal_blocks = np.split(society_all, 5, axis=1)

print("Five Equal Blocks:")
for i, block in enumerate(equal_blocks, 1):
    print("Block", i, "Shape:", block.shape)

# Split into 4 uneven blocks
uneven_blocks = np.array_split(society_all, 4, axis=1)

print("\nFour Uneven Blocks:")
for i, block in enumerate(uneven_blocks, 1):
    print("Block", i, "Shape:", block.shape)

assert len(equal_blocks) == 5
assert all(block.shape == (12, 7) for block in equal_blocks)
assert uneven_blocks[-1].shape == (12, 8)

print("\nAll splitting checks passed!")
```

Task 4: Delete, insert, append, copy and view

```python

# Create a safe copy
society_edit = society_all.copy()

# Remove the 5th flat column
society_edit = np.delete(society_edit, 4, axis=1)

# Insert a new flat with 20 KL for every month
society_edit = np.insert(
    society_edit, 4, 20, axis=1
)

# Append a 13th month by copying March
society_edit = np.append(
    society_edit,
    society_edit[-1:, :],
    axis=0
)

print("Edited Society Shape:", society_edit.shape)

assert society_edit.shape == (13, 35)

# Check original data is unchanged
np.random.seed(42)
original_water = np.random.randint(8, 41, size=(12, 20))

np.random.seed(7)
original_b = np.random.randint(8, 41, size=(12, 15))

original_society = np.hstack((original_water, original_b))

assert np.array_equal(society_all, original_society)

print("Original Society Unchanged: True")

# Demonstrate view
peek = society_all.view()

peek[0, 0] = 99

assert society_all[0, 0] == 99

print("Updated Original First Value:", society_all[0, 0])
print("View Check Passed!")
```
