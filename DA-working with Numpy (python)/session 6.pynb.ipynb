# Housing Society Water Usage Analysis
## NumPy Session 6 – Indexing & Slicing

Analysis of monthly water consumption for 20 flats over 12 months.

Task 1: Generate water usage data

```python

import numpy as np

np.random.seed(42)

water_kl = np.random.randint(8, 41, size=(12, 20))

flat_no = np.arange(101, 121)

months = np.array([
    "April", "May", "June", "July",
    "August", "September", "October",
    "November", "December", "January",
    "February", "March"
])

print("Water Data Shape:", water_kl.shape)

print("March Reading of Flat 120:", water_kl[-1, -1])

print("April Readings:")
print(water_kl[0, :])

print("Full Year Readings of Flat 105:")
print(water_kl[:, 4])
```

Task 2: Slicing arrays

```python

monsoon = water_kl[3:7, :]

first_quarter = water_kl[0:3, 0:10]

alternate_months = water_kl[::2, :]

reversed_flats = water_kl[:, ::-1]

print("Monsoon Shape:", monsoon.shape)
print("First Quarter Shape:", first_quarter.shape)
print("Alternate Months Shape:", alternate_months.shape)
print("Reversed Flats Shape:", reversed_flats.shape)

assert monsoon.shape == (4, 20)
assert first_quarter.shape == (3, 10)
assert alternate_months.shape == (6, 20)
assert reversed_flats.shape == (12, 20)

print("All slicing checks passed!")
```

Task 3: Fancy indexing

```python

audit_month_idx = [0, 5, 11]

audit_flat_idx = [2, 9, 14, 19]

audit_sample = water_kl[audit_month_idx][:, audit_flat_idx]

print("Audit Sample:")
print(audit_sample)

print("Selected Months:")
print(months[audit_month_idx])

print("Selected Flats:")
print(flat_no[audit_flat_idx])

print("Audit Sample Shape:", audit_sample.shape)

assert audit_sample.shape == (3, 4)

print("Fancy Indexing Check Passed!")
```

Task 4: Boolean indexing and np.where()

```python

heavy = water_kl > 35

print("Heavy Usage Readings:")
print(water_kl[heavy])

print("Total Heavy Usage Readings:", heavy.sum())

row_idx, col_idx = np.where(heavy)

print("\nHeavy Usage Details:")

for r, c in zip(row_idx, col_idx):
    print(
        "Month:", months[r],
        "| Flat:", flat_no[c],
        "| Water:", water_kl[r, c], "KL"
    )

capped_kl = np.where(water_kl > 35, 35, water_kl)

print("\nMaximum Capped Water Usage:", capped_kl.max())

assert capped_kl.max() == 35

print("Boolean Indexing Check Passed!")
```
