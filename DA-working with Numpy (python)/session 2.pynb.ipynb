 – NumPy Session 2
Creating arrays, student marks, attendance, fees and random data.

Task 1: Create student data

```python

import numpy as np

np.random.seed(42)

roll_no = np.arange(101, 141)

marks = np.random.randint(35, 101, size=(40, 5))

attendance_pct = np.random.randint(60, 101, size=40)

batch = np.random.choice(
    ["Morning", "Evening", "Weekend"],
    size=40,
    p=[0.5, 0.3, 0.2]
)

print("Roll Numbers Shape:", roll_no.shape)
print("Marks Shape:", marks.shape)
print("Attendance Shape:", attendance_pct.shape)
print("Batch Shape:", batch.shape)
```

Task 2: Create blank register templates

```python

pass_mark_grid = np.ones((40, 5)) * 35

grace_marks = np.zeros(40)

scratch = np.empty((40, 5))

subject_picker = np.eye(5)

subject_ids = np.identity(5)

print("Pass Mark Grid:", pass_mark_grid.shape)
print("Grace Marks:", grace_marks.shape)
print("Scratch:", scratch.shape)
print("Subject Picker:", subject_picker.shape)
print("Subject IDs:", subject_ids.shape)
```

Task 3: Create fee arrays

```python

fee_slabs = np.linspace(1000, 5000, 5)

late_fee_ladder = np.logspace(0, 4, 5, base=2)

fee_inr = np.random.choice(fee_slabs, size=40)

print("Fee Slabs:", fee_slabs)
print("Late Fee Ladder:", late_fee_ladder)
print("Fee INR Shape:", fee_inr.shape)
```

Task 4: Retest marks and lucky draw

```python

np.random.seed(7)

retest_marks = marks + np.random.randn(40, 5) * 3

lucky_draw = np.random.rand(40)

print("Marks Shape:", marks.shape)
print("Retest Marks Shape:", retest_marks.shape)
print("Lucky Draw Shape:", lucky_draw.shape)

assert marks.shape == (40, 5)
assert retest_marks.shape == (40, 5)
assert lucky_draw.shape == (40,)

print("All shape checks passed!")
```
