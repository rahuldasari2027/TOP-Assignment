Step 2 – Task 1: Create 200,000 prices

```python

import numpy as np

prices_list = [5 + (i * 37) % 496 for i in range(200000)]

prices_arr = np.array(prices_list)

print("List length:", len(prices_list))
print("Array size:", prices_arr.size)
print("Array total:", prices_arr.sum())
```

Step 3 – Task 2: Compare Python list and NumPy speed

```python

import time

list_times = []
array_times = []

for i in range(3):
    start = time.perf_counter()
    gst_list = [p * 1.18 for p in prices_list]
    list_times.append(time.perf_counter() - start)

    start = time.perf_counter()
    gst_arr = prices_arr * 1.18
    array_times.append(time.perf_counter() - start)

list_ms = min(list_times) * 1000
array_ms = min(array_times) * 1000
speedup = list_ms / array_ms

print(f"List time: {list_ms:.3f} ms")
print(f"NumPy time: {array_ms:.3f} ms")
print(f"Speed-up: {speedup:.1f}x")
```

Step 4 – Task 3: Compare memory usage

```python

import sys

list_bytes = sys.getsizeof(prices_list) + sum(
    sys.getsizeof(p) for p in prices_list
)

array_bytes = prices_arr.nbytes

prices_small = np.array(prices_list, dtype=np.int16)
small_bytes = prices_small.nbytes

print("Python list bytes:", list_bytes)
print("NumPy array bytes:", array_bytes)
print("Small array bytes:", small_bytes)
print(f"List is {list_bytes / small_bytes:.1f} times bigger")
```

Step 5 – Task 4: Calculate GST and festival discount

```python

total_gst = (prices_arr * 0.18).sum()
discount_total = (prices_arr * 0.95).sum()

list_gst = sum(p * 0.18 for p in prices_list)
list_discount = sum(p * 0.95 for p in prices_list)

print("Total GST (18%):", total_gst)
print("Total after 5% discount:", discount_total)

print("GST check:", abs(total_gst - list_gst) < 0.01)
print("Discount check:", abs(discount_total - list_discount) < 0.01)
```
