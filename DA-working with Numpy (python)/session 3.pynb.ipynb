 – Array Attributes & Properties
NumPy Session 3 Assignment

```python

import numpy as np

np.random.seed(42)

marks = np.random.randint(35, 101, size=(40, 5))

np.random.seed(7)
retest_marks = marks + np.random.randn(40, 5) * 3

fee_inr = np.random.choice(
    np.linspace(1000, 5000, 5),
    size=40
)

arrays = {
    "marks": marks,
    "retest_marks": retest_marks,
    "fee_inr": fee_inr
}

print("Array Attributes")
print("-" * 85)
print(f"{'Array':<15} {'ndim':<6} {'shape':<12} {'size':<7} {'dtype':<10} {'itemsize':<10} {'nbytes':<10}")

for name, arr in arrays.items():
    print(f"{name:<15} {arr.ndim:<6} {str(arr.shape):<12} {arr.size:<7} {str(arr.dtype):<10} {arr.itemsize:<10} {arr.nbytes:<10}")
    assert arr.nbytes == arr.size * arr.itemsize

print("All attribute checks passed!")
```

Task 2: Reshape, ravel and flatten

```python

marks_batches = marks.reshape(4, 10, 5)

marks_ravel = marks.ravel()
marks_flat = marks.flatten()

print("Marks Batches Shape:", marks_batches.shape)
print("Marks Ravel Shape:", marks_ravel.shape)
print("Marks Flatten Shape:", marks_flat.shape)

original_value = marks[0, 0]

marks_ravel[0] = 0

assert marks[0, 0] == 0
print("Ravel changed original:", marks[0, 0] == 0)

marks_flat[0] = 999

assert marks[0, 0] == 0
print("Flatten did not change original:", marks[0, 0] == 0)

marks[0, 0] = original_value

print("Original value restored:", marks[0, 0])
```

Task 3: Transpose arrays

```python

marks_by_subject = marks.T

marks_subject_batch = np.transpose(
    marks_batches, (2, 0, 1)
)

print("Marks By Subject:")
print("Dimensions:", marks_by_subject.ndim)
print("Shape:", marks_by_subject.shape)

print("\nMarks Subject Batch:")
print("Dimensions:", marks_subject_batch.ndim)
print("Shape:", marks_subject_batch.shape)

assert marks_by_subject.shape == (5, 40)
assert marks_subject_batch.shape == (5, 4, 10)

print("\nAll transpose checks passed!")
```

Task 4: Resize arrays

```python

seat_fees = np.resize(fee_inr, (45,))

fee_copy = fee_inr.copy()
fee_copy.resize(45, refcheck=False)

print("Seat Fees Size:", seat_fees.size)
print("Seat Fees Bytes:", seat_fees.nbytes)

print("Fee Copy Size:", fee_copy.size)
print("Fee Copy Bytes:", fee_copy.nbytes)

print("\nExtra 5 Seat Fees:", seat_fees[40:])
print("Extra 5 Fee Copy:", fee_copy[40:])

assert np.array_equal(seat_fees[40:], fee_inr[:5])
assert np.all(fee_copy[40:] == 0)

print("\nAll resize checks passed!")
```
