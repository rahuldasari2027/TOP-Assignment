# Mobile Recharge Usage Analysis
## Seaborn Session 10 – Advanced Styling

This project analyzes 300 fictional prepaid mobile users and compares Seaborn palettes, contexts and corporate chart styling.

Task 1: Generate 300 prepaid users

```python

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate 300 user IDs
user_id = np.arange(1, 301)

# Circle
circle = np.random.choice(
    ["Gujarat", "Maharashtra", "Rajasthan"],
    size=300
)

# Recharge plans
plan = np.random.choice(
    ["Rs199", "Rs299", "Rs599"],
    size=300,
    p=[0.5, 0.3, 0.2]
)

# Average data usage by plan
data_mean = np.where(
    plan == "Rs199", 18,
    np.where(plan == "Rs299", 32, 60)
)

# Generate data usage
data_gb_used = np.random.normal(
    loc=data_mean,
    scale=data_mean * 0.25
)

data_gb_used = np.round(
    np.maximum(data_gb_used, 1),
    1
)

# Generate calling minutes
calls_min = np.random.normal(
    loc=420,
    scale=120,
    size=300
)

calls_min = np.round(
    np.maximum(calls_min, 30)
).astype(int)

# Create DataFrame
df = pd.DataFrame({
    "user_id": user_id,
    "circle": circle,
    "plan": plan,
    "data_gb_used": data_gb_used,
    "calls_min": calls_min
})

# Save CSV
df.to_csv("recharge_usage.csv", index=False)

print(df.head())
print("Total Users:", len(df))

assert len(df) == 300

# Basic barplot
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="plan",
    y="data_gb_used",
    hue="circle",
    errorbar=None
)

plt.title("Average Data Usage by Recharge Plan")
plt.xlabel("Recharge Plan")
plt.ylabel("Average Data Usage (GB)")

plt.tight_layout()
plt.savefig("recharge_before.png", dpi=150)
plt.show()

print("Task 1 Completed!")
```

Task 2: Compare Seaborn palettes

```python

palettes = ["deep", "muted", "colorblind"]

fig, axes = plt.subplots(
    1, 3,
    figsize=(18, 5)
)

for ax, palette_name in zip(axes, palettes):

    sns.barplot(
        data=df,
        x="plan",
        y="data_gb_used",
        hue="circle",
        palette=palette_name,
        errorbar=None,
        ax=ax
    )

    ax.set_title(palette_name)
    ax.set_xlabel("Recharge Plan")
    ax.set_ylabel("Average Data Usage (GB)")

plt.tight_layout()

plt.savefig(
    "recharge_palettes.png",
    dpi=150
)

plt.show()

print("Task 2 Completed!")
```

Task 3: Compare notebook, talk and poster contexts

```python

contexts = ["notebook", "talk", "poster"]

for context_name in contexts:

    # Set context before creating figure
    sns.set_context(context_name)

    plt.figure(figsize=(8, 5))

    sns.barplot(
        data=df,
        x="plan",
        y="data_gb_used",
        hue="circle",
        errorbar=None
    )

    plt.title(
        f"Recharge Usage - {context_name}"
    )

    plt.xlabel("Recharge Plan")
    plt.ylabel("Average Data Usage (GB)")

    plt.tight_layout()

    plt.savefig(
        f"recharge_{context_name}.png",
        dpi=150
    )

    plt.show()

    print(context_name, "chart saved!")

# Reset Seaborn context
sns.set_theme()
print("Task 3 Completed!")
```

Task 4: Create corporate-style Seaborn theme

```python

# Create corporate theme function
def apply_corporate_theme():

    sns.set_theme(
        style="whitegrid",
        context="talk",
        palette="colorblind",
        font_scale=0.9,
        rc={
            "axes.spines.top": False,
            "axes.spines.right": False
        }
    )

# Apply corporate theme
apply_corporate_theme()

# Create final chart
fig, ax = plt.subplots(figsize=(11, 6))

sns.barplot(
    data=df,
    x="plan",
    y="data_gb_used",
    hue="circle",
    errorbar=None,
    ax=ax
)

ax.set_title(
    "Corporate Report - Mobile Recharge Usage"
)

ax.set_xlabel("Recharge Plan")
ax.set_ylabel("Average Data Usage (GB)")

ax.legend(title="Circle")

plt.tight_layout()

plt.savefig(
    "recharge_corporate.png",
    dpi=150
)

plt.show()

print("Task 4 Completed!")
print("Session 10 Completed Successfully!")
```
