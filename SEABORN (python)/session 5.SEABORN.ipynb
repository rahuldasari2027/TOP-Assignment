# Penguin Dataset – Multivariate Analysis
## Seaborn Session 5 – Pairplot

This project explores relationships between penguin measurements using Seaborn pairplots, species, sex and KDE distributions.

Task 1: Load penguins dataset and create a basic pairplot

```python

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = sns.load_dataset("penguins")

# Remove missing values
df = df.dropna().copy()

# Save cleaned data
df.to_csv("penguins_clean.csv", index=False)

print(df.head())
print("Total Rows:", len(df))

assert len(df) == 333

numeric_cols = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]

# Basic pairplot
g = sns.pairplot(
    data=df,
    vars=numeric_cols
)

g.savefig("penguins_pairplot_basic.png", dpi=150)
# Pairplot with histogram diagonal
g1 = sns.pairplot(
    data=df,
    vars=numeric_cols,
    hue="species",
    diag_kind="hist"
)

g1.savefig("penguins_diag_hist.png", dpi=150)
plt.show()

print("Histogram pairplot saved!")

# Pairplot with KDE diagonal
g2 = sns.pairplot(
    data=df,
    vars=numeric_cols,
    hue="species",
    diag_kind="kde"
)

g2.savefig("penguins_diag_kde.png", dpi=150)
plt.show()

print("KDE pairplot saved!")

plt.show()

print("Basic pairplot saved!")
```

Task 2: Pairplot with species

```python

# Pairplot with histogram diagonal
g1 = sns.pairplot(
    data=df,
    vars=numeric_cols,
    hue="species",
    diag_kind="hist"
)

g1.savefig("penguins_diag_hist.png", dpi=150)
plt.show()

print("Histogram pairplot saved!")

# Pairplot with KDE diagonal
g2 = sns.pairplot(
    data=df,
    vars=numeric_cols,
    hue="species",
    diag_kind="kde"
)

g2.savefig("penguins_diag_kde.png", dpi=150)
plt.show()

print("KDE pairplot saved!")
```

Task 3: Pairplot with sex and lower triangle

```python

g3 = sns.pairplot(
    data=df,
    vars=numeric_cols,
    hue="sex",
    corner=True,
    height=2,
    plot_kws={
        "alpha": 0.6,
        "s": 25
    }
)

g3.savefig("penguins_sex_corner.png", dpi=150)
plt.show()

print("Sex corner pairplot saved!")
```

Task 4: Final pairplot and species summary

```python

final_cols = [
    "bill_length_mm",
    "flipper_length_mm",
    "body_mass_g"
]

# Final pairplot
g4 = sns.pairplot(
    data=df,
    vars=final_cols,
    hue="species",
    diag_kind="kde"
)

g4.figure.suptitle(
    "Penguin Measurements by Species",
    y=1.02
)

g4.savefig("penguins_final_pairplot.png", dpi=150)
plt.show()

# Species summary
species_summary = (
    df.groupby("species")[final_cols]
    .mean()
    .round(2)
)

print("PENGUIN SPECIES SUMMARY")
print(species_summary)

print("\nSession 5 completed successfully!")
```
