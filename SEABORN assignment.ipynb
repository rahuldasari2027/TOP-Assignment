```python

# M10-A1 Section D
# AI-Generated Original Code

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Generate synthetic food delivery dataset
np.random.seed(42)

n = 200

df = pd.DataFrame({
    "order_value": np.random.uniform(100, 800, n),
    "distance_km": np.random.uniform(1, 20, n),
    "delivery_time_mins": np.random.normal(30, 8, n),
    "rating": np.random.uniform(1, 5, n),
    "discount_pct": np.random.uniform(0, 30, n),
    "cuisine_type": np.random.choice(
        ["Indian", "Chinese", "Fast Food"],
        n
    )
})

print("FOOD DELIVERY DATASET")
print(df.head())

print("\nDataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

```python

# Section D - Seaborn Pairplot

print("\nCreating Seaborn Pairplot...")

pair = sns.pairplot(
    data=df,
    vars=[
        "order_value",
        "distance_km",
        "delivery_time_mins",
        "rating",
        "discount_pct"
    ],
    hue="cuisine_type",
    diag_kind="hist"
)

pair.fig.suptitle(
    "Food Delivery Analytics - Pairplot",
    y=1.02
)

# Save pairplot as PNG
pair.savefig(
    "D:/M10_A1_Section_D/food_delivery_pairplot.png"
)

plt.close(pair.fig)

print("Pairplot saved successfully!")
```

```python

# Section D - Pearson Correlation Heatmap

print("\nCreating Correlation Heatmap...")

# Select numeric columns
numeric_df = df.select_dtypes(include="number")

# Calculate Pearson correlation
correlation = numeric_df.corr(method="pearson")

print("\nPEARSON CORRELATION MATRIX")
print(correlation.round(2))

# Create heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Food Delivery Correlation Heatmap")
plt.tight_layout()

# Save heatmap at 200 DPI
plt.savefig(
    "D:/M10_A1_Section_D/food_delivery_correlation_heatmap.png",
    dpi=200,
    bbox_inches="tight"
)

plt.close()

print("\nHeatmap saved successfully!")
print("Section D original code completed!")
```
