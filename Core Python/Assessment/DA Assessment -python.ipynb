Healthcare Industry

```python
class ClinicAppointment:
    def __init__(self):
        self.appointments = []

    def view(self):
        mobile = "9714788103"
        for a in self.appointments:
            if a["mobile"] == mobile:
                print(a)
                return
        print("Not found")


c = ClinicAppointment()

# Add test data
c.appointments.append({
    "name": "Rahul",
    "mobile": "9714788103",
    "doctor": "Dr kaushal",
    "slot": "10am"
})

# View test
c.view()
```

School Management System

```python
class SchoolManagement:
    def __init__(self):
        self.students = {}
        self.student_id = 1

    def new_admission(self, name, age, student_class, mobile):
        if age < 5 or age > 18:
            print("Invalid age")
            return

        if not (mobile.isdigit() and len(mobile) == 10):
            print("Invalid mobile")
            return

        sid = self.student_id
        self.students[sid] = {
            "name": name,
            "age": age,
            "class": student_class,
            "mobile": mobile
        }
        self.student_id += 1
        print("Added:", sid)

    def view_student(self, sid):
        print(self.students.get(sid, "Not found"))


# Testing without input()
obj = SchoolManagement()
obj.new_admission("Rahul", 15, 10, "9876543210")
obj.view_student(1)
```

Transport Reservation System (Bus Ticketing)

```python
def book_ticket(self):
    # Hardcoded values instead of input()
    name = "Rahul"
    age = 22
    mobile = "9876543210"
    route_choice = 1

    if len(mobile) != 10 or not mobile.isdigit():
        print("Invalid mobile number")
        return

    if route_choice not in self.routes:
        print("Invalid route")
        return

    if self.seats[route_choice] >= 40:
        print("Bus Full")
        return

    self.seats[route_choice] += 1
    seat_number = self.seats[route_choice]

    ticket = {
        "name": name,
        "age": age,
        "mobile": mobile,
        "route": self.routes[route_choice][0],
        "price": self.routes[route_choice][1],
        "seat": seat_number
    }

    self.tickets[self.ticket_id] = ticket
    print("Ticket Booked! ID:", self.ticket_id)

    self.ticket_id += 1
```
