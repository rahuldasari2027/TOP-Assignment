# Ahmedabad Auto-Rickshaw Trip Analysis
NumPy Session 4 – Array Operations

This project analyzes 500 auto-rickshaw trips using NumPy arithmetic, logical operations and broadcasting.

```python

import numpy as np

np.random.seed(42)

trip_id = np.arange(1, 501)

distance_km = np.random.rand(500) * 24 + 1

wait_min = np.random.randint(0, 16, 500)

hour = np.random.randint(6, 24, 500)

surge_level = np.random.randint(0, 4, 500)

print("Trip ID Shape:", trip_id.shape)
print("Distance Shape:", distance_km.shape)
print("Wait Time Shape:", wait_min.shape)
print("Hour Shape:", hour.shape)
print("Surge Level Shape:", surge_level.shape)
```

Task 2: Calculate fare and driver pay

```python

fare_inr = (20 + 12 * distance_km + 1.5 * wait_min) * (1.1 ** surge_level)

driver_pay = fare_inr - fare_inr / 5

zone = distance_km // 5

km_in_zone = distance_km % 5

print("First 5 Fares:", fare_inr[:5])
print("First 5 Driver Payments:", driver_pay[:5])
print("First 5 Zones:", zone[:5])
print("First 5 KM in Zone:", km_in_zone[:5])
```
