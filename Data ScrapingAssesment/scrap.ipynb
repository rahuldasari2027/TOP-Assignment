11.7_Scrapping_using_csv.ipynb

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', None, 'David', 'Eve'],
    'age': ['25', '30', None, '45', '40', ''],
    'date_joined': ['2023-01-01', '2022-12-31', '2023-01-15', None, '2023-01-20', '2023-01-25'],
    'email': ['alice@example.com', 'bob@example', 'charlie@example.com', None, 'david@example.com', 'eve@example.com']
})

print("Original Data:\n", data)

# Handle missing values
data['name'] = data['name'].fillna('Unknown')

data['age'] = pd.to_numeric(data['age'], errors='coerce')
data['age'] = data['age'].fillna(data['age'].median())

# ✅ Version-safe forward fill
try:
    data['date_joined'] = data['date_joined'].ffill()
except:
    data['date_joined'] = data['date_joined'].fillna(method='ffill')

# Email validation
data['email'] = data['email'].apply(
    lambda x: x if isinstance(x, str) and '@' in x else 'unknown@example.com'
)

print("\nAfter Cleaning:\n", data)

# Format date
data['date_joined'] = pd.to_datetime(data['date_joined'])
data['date_joined'] = data['date_joined'].dt.strftime('%Y-%m-%d')

print("\nFinal Data:\n", data)
```

```python
import requests
from bs4 import BeautifulSoup

url = 'https://en.wikipedia.org/wiki/List_of_countries_by_GDP_(nominal)'

# Add headers (VERY IMPORTANT for Wikipedia)
headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, 'html.parser')

# Find ALL wikitable tables
tables = soup.find_all('table', class_='wikitable')

if not tables:
    print("No tables found ❌")
else:
    table = tables[0]  # first GDP table
    rows = table.find_all('tr')

    for row in rows[1:11]:
        cols = row.find_all('td')
        
        if len(cols) >= 2:
            country = cols[0].text.strip()
            gdp = cols[1].text.strip()
            
            print(f"Country: {country}")
            print(f"GDP: {gdp}")
            print("---")
```
